# Notebook 04: Model Evaluation & Benchmark Comparison

## Objective
Systematically compare classical baselines (MultinomialNB, Logistic Regression, Linear SVC) and Fine-Tuned Transformer architectures (DistilBERT, Context DistilBERT, RoBERTa) on the MELD test set.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
print("Evaluation libraries loaded successfully!")
```

```python
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
metrics_file = PROJECT_ROOT / "results" / "metrics" / "model_comparison.csv"
df_metrics = pd.read_csv(metrics_file)
print("Benchmark comparison table:")
df_metrics
```

```python
plt.figure(figsize=(10, 5))
palette = ["#8fa3bf", "#8fa3bf", "#8fa3bf", "#315a7d", "#1e3d59", "#172b4d"]
bars = plt.bar(df_metrics["Model"], df_metrics["Accuracy (%)"], color=palette)
plt.xticks(rotation=20, ha="right")
plt.ylabel("Accuracy (%)", fontweight="bold")
plt.title("Sentiment Model Accuracy on MELD Benchmark", fontweight="bold")
for b in bars:
    h = b.get_height()
    plt.text(b.get_x() + b.get_width()/2, h + 0.6, f"{h:.1f}%", ha="center", fontweight="bold")
plt.ylim(40, 75)
plt.tight_layout()
plt.show()
```

### Conclusion & Observations
1. **Transformer Advantage:** Fine-tuned DistilBERT and RoBERTa outperform classical models by over 17% in accuracy.
2. **Context Impact:** Incorporating 2 preceding dialogue turns elevates Macro F1 from 65.45% to 66.94%.
