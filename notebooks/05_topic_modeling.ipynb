# Notebook 05: Conversational Topic Modeling

## Objective
Discover latent themes and topics across conversational turns in MELD using TF-IDF and Non-negative Matrix Factorization (NMF).

```python
import pandas as pd
from pathlib import Path
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF

print("Libraries imported successfully!")
```

```python
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
train_df = pd.read_csv(PROJECT_ROOT / "data" / "processed" / "train_processed.csv")

vec = TfidfVectorizer(stop_words="english", max_features=1000, ngram_range=(1, 2))
X = vec.fit_transform(train_df["clean_text"].dropna())

n_topics = 4
nmf = NMF(n_components=n_topics, random_state=42, init="nndsvda", max_iter=300)
W = nmf.fit_transform(X)
words = vec.get_feature_names_out()

print("Top terms by topic:")
for i, comp in enumerate(nmf.components_):
    top = [words[idx] for idx in comp.argsort()[-6:][::-1]]
    print(f"Topic {i+1}: {', '.join(top)}")
```
